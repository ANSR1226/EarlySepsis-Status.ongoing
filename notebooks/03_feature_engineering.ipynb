DATASET AGGREGATION

```python
import pandas as pd
```

```python
setA = r'data\interim\part_processA.csv'
setB = r'data\interim\part_processB.csv'
```

```python
dfA = pd.read_csv(setA)
dfB = pd.read_csv(setB)

dfA.drop(columns = ['Unnamed: 0'], inplace = True)
dfB.drop(columns = ['Unnamed: 0'], inplace = True)
```

```python
dfB.info()
```

```python
vital_signs = ['HR', 'O2Sat', 'Temp', 'SBP', 'MAP', 'DBP', 'Resp']
metabolic = ['BaseExcess', 'HCO3', 'Lactate', 'Glucose']
kidney_function = ['BUN', 'Creatinine', 'Calcium', 'Chloride', 'Magnesium', 'Phosphate', 'Potassium']
hematology = ['Hct', 'Hgb', 'WBC', 'Platelets', 'PTT', 'Fibrinogen']
blood_gas_function = ['FiO2', 'pH', 'PaCO2']
demographics = ['Age', 'Gender', 'HospAdmTime', 'Unit1', 'Unit2']
time_label = ['ICULOS', 'SepsisLabel', 'patient_id']
```

```python
agg_col = vital_signs + metabolic + kidney_function + hematology + blood_gas_function

agg_dict = {col : ['mean', 'std', 'max', 'min', 'count'] for col in agg_col
       }
agg.update({col : ['first'] for col in demographics})
agg.update({'ICULOS' : ['max']})
agg.update({'SepsisLabel' : ['max']})
```

```python
agg_dict
```

```python
dfA_final = dfA.groupby('patient_id').agg(agg_dict)
```

```python
dfA_final.info()
```

```python
dfA_final.columns = ['_'.join(col) for col in dfA_final.columns] #flattening the name

# the old name format is the nested column form that comes out after groupby().agg(), like:

# ('HR', 'mean')

# The new one is the flattened version, like:

# HR_mean
```

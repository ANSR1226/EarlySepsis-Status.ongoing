#### Core vital signs (most important for sepsis)
###### HR (Heart Rate): Beats per minute. Normal 60–100. Tachycardia (high HR) is an early sign of sepsis.

###### O2Sat (Oxygen Saturation): Percentage of oxygen in the blood from pulse oximeter. Normal 95–100%. Low values indicate hypoxia and lung dysfunction.

###### Temp (Temperature): Body temperature in °C. Normal 36.5–37.5. Fever or abnormally low temperature both suggest infection.

###### Resp (Respiratory Rate): Breaths per minute. Normal 12–20. High respiratory rate (tachypnea) is an early sepsis warning.

###### SaO2 (Arterial Oxygen Saturation): Oxygen saturation from blood gas. Normal 95–100%. Confirms oxygenation status.

###### FiO2 (Fraction of Inspired Oxygen): Percentage of oxygen in inhaled air. Room air is ~21%. Higher values mean the patient needs supplemental oxygen.

###### EtCO2 (End-Tidal CO₂): Partial pressure of CO₂ at the end of exhalation (mmHg). Normal 35–45. Reflects ventilation and perfusion.

###### Blood pressure (circulation and perfusion)
SBP (Systolic Blood Pressure): Top number of blood pressure (mmHg). Normal 90–120. Low SBP can indicate shock.

###### MAP (Mean Arterial Pressure): Average arterial pressure (mmHg). Normal 70–100. MAP < 65 suggests organ hypoperfusion; this is a key sepsis marker.

###### DBP (Diastolic Blood Pressure): Bottom number of blood pressure (mmHg). Normal 60–80. Low DBP can occur in shock.

#### Acid-base and ventilation status
###### pH: Blood pH. Normal 7.35–7.45. Lower pH (acidemia) can occur in sepsis due to poor perfusion.

###### PaCO2: Partial pressure of carbon dioxide in arterial blood (mmHg). Normal 35–45. High values indicate respiratory acidosis.

###### BaseExcess: Base excess (mmol/L). Normal −2 to +2. Negative values indicate metabolic acidosis (often from lactate).

###### HCO3: Bicarbonate concentration (mmol/L). Normal 22–28. Low bicarbonate suggests metabolic acidosis.

#### Metabolic, kidney, and liver markers
###### Lactate: Lactate level (mmol/L). Normal 0.5–2.0. Elevated lactate is one of the strongest markers of tissue hypoxia and sepsis; clinically ≥2 is concerning.

###### Glucose: Blood glucose (mg/dL). Normal ~70–100 fasting. Glucose dysregulation is common in sepsis.

###### Creatinine: Serum creatinine (mg/dL). Normal ~0.7–1.3. Elevated creatinine indicates kidney dysfunction.

###### BUN: Blood Urea Nitrogen (mg/dL). Normal ~7–20. High BUN suggests kidney impairment.

###### AST: Aspartate aminotransferase (U/L). Normal ~10–40. Elevated AST can indicate liver stress.

###### Bilirubin_direct: Direct (conjugated) bilirubin (mg/dL). Normal <0.3. Elevated values suggest liver dysfunction.

###### Bilirubin_total: Total bilirubin (mg/dL). Normal <1.2. High values indicate liver dysfunction or hemolysis.

###### Alkalinephos: Alkaline phosphatase (U/L). Normal ~44–147. Elevated in liver or bone disease.

###### Calcium: Serum calcium (mg/dL). Normal ~8.5–10.5. Abnormal calcium can occur in critical illness.

###### Magnesium: Serum magnesium (mg/dL). Normal ~1.7–2.2. Electrolyte imbalance in sepsis.

###### Phosphate: Serum phosphate (mg/dL). Normal ~2.5–4.5. Electrolyte disturbance possible in sepsis.

###### Chloride: Serum chloride (mEq/L). Normal ~96–106. Part of acid-base balance.

###### Potassium: Serum potassium (mEq/L). Normal ~3.5–5.0. Critical electrolyte; abnormalities can be life-threatening.

###### TroponinI: Troponin I (ng/mL). Normal <0.04. Elevated troponin indicates heart muscle injury.


### Merging Training Data A

```python
import os

folderA_directory = r'data\raw\training_setA'
fileA_directories = os.listdir(folderA_directory)

dfs_A = []

for fname in fileA_directories:
    if fname.endswith('.psv'):
        path = os.path.join(folderA_directory, fname)
        df_A = pd.read_csv(path, sep='|')
        df_A['patient_id'] = fname
        dfs_A.append(df_A)
```

```python
dfs_A
```

```python
combined_dfA = pd.concat(dfs_A, ignore_index = True)
```
